This script performs the following steps to prepare the dataset:

1. Importing Datasets : Loads all CSV files required for the analysis.

2. Dataset Exploration : Reviews and summarizes each CSV file to understand the structure,columns, and overall data content.

# **📥 Setup & Configuration**

```python
# 📘 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np

# 📂 2. LOAD DATA
customers = pd.read_csv('olist_customers_dataset.csv')
orders = pd.read_csv('olist_orders_dataset.csv')
order_items = pd.read_csv('olist_order_items_dataset.csv')
products = pd.read_csv('olist_products_dataset.csv')
payments = pd.read_csv('olist_order_payments_dataset.csv')
reviews = pd.read_csv('olist_order_reviews_dataset.csv')
sellers = pd.read_csv('olist_sellers_dataset.csv')
geolocation = pd.read_csv('olist_geolocation_dataset.csv')
category_translation = pd.read_csv('product_category_name_translation.csv')
```

# *🧹 Data Cleaning & Wrangling*

```python
# ====================== BASIC CLEANING FUNCTION ======================
def clean_table(df, name):
    print(f"\n{'='*60}")
    print(f"Cleaning: {name}")
    print(f"Original Shape: {df.shape}")
    
    # 1. Drop exact duplicate rows (if any)
    df = df.drop_duplicates()
    
    # 2. Strip whitespace from all string columns
    for col in df.select_dtypes(include='object').columns:
        df[col] = df[col].str.strip()
    
    print(f"Shape after cleaning: {df.shape}")
    return df
```

```python
# Apply basic cleaning
customers = clean_table(customers, "Customers")
orders = clean_table(orders, "Orders")
order_items = clean_table(order_items, "Order Items")
products = clean_table(products, "Products")
payments = clean_table(payments, "Payments")
reviews = clean_table(reviews, "Reviews")
```

```python
# ====================== SPECIFIC CLEANING ======================

# 1. Customers Table

customers['customer_city'] = customers['customer_city'].str.lower()
customers['customer_state'] = customers['customer_state'].str.upper()
print('Customer Table Done ')

# 2. Orders Table 

orders['order_status'] = orders['order_status'].str.lower()

date_columns = ['order_purchase_timestamp', 'order_approved_at', 
                'order_delivered_carrier_date', 'order_delivered_customer_date', 
                'order_estimated_delivery_date']

for col in date_columns:
    if col in orders.columns:
        orders[col]=pd.to_datetime(orders[col], errors='coerce')
# Creating new columns
orders['delivery_days'] = (orders['order_delivered_customer_date']-orders['order_purchase_timestamp']).dt.days
orders['is_late'] = orders['order_delivered_customer_date'] > orders['order_estimated_delivery_date']


# 3. Order Items
order_items['total_value'] = order_items['price'] + order_items['freight_value']


# 4. Products Table 

products['product_category_name'] = products['product_category_name'].str.lower()

if category_translation is not None:
    products = products.merge(category_translation, 
                              on='product_category_name', 
                              how='left')

# 5. Payments
payments['payment_type'] = payments['payment_type'].str.title()

# 6. Reviews
reviews['has_review'] = reviews['review_score'].notna().astype(int)

# ====================== FINAL SUMMARY ======================
print("\n" + "="*80)
print("DATA CLEANING COMPLETED!")
print("="*80)
print(f"Customers Shape     : {customers.shape}")
print(f"Orders Shape        : {orders.shape}")
print(f"Order Items Shape   : {order_items.shape}")
print(f"Products Shape      : {products.shape}")
print(f"Payments Shape      : {payments.shape}")
print(f"Reviews Shape       : {reviews.shape}")

# Check remaining missing values in key columns
print("\nMissing Values in Orders (Key Columns):")
print(orders[['order_delivered_customer_date', 'delivery_days']].isnull().sum())

```

#  🔄 Dataset Merging & Saving As CSV File

```python
# ====================== MERGING TABLES ======================

# Orders + Customers
master = orders.merge(customers , on = 'customer_id', how = 'left' )

# Order Items 
master = master.merge(order_items, on='order_id', how ='left')

# Products 
master = master.merge(products[['product_id', 'product_category_name', 'product_category_name_english']], 
                      on='product_id', how='left')

#Payments
master = master.merge(payments, on='order_id', how='left')

# Reviews
master = master.merge(reviews[['order_id', 'review_score', 'has_review']], 
                      on='order_id', how='left')

print("✅ Master table created successfully!")
print(f"Final Master Shape: {master.shape}")


# ====================== SUMMARY OF MASTER TABLE ======================
print("\n" + "="*80)
print("MASTER TABLE SUMMARY")
print("="*80)
print(master.info())



print("\nMissing Values in Master Table (Top 10 columns with most missing):")
print(master.isnull().sum().sort_values(ascending=False).head(12))


# Saving the master file
master.to_csv('master_sales_data.csv', index=False)
print("\n✅ Master file saved as 'master_sales_data.csv'")
```
